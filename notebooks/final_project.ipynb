### Откуда брать задачу

Хорошая задача выглядит так: есть человек, который сейчас тратит на это время, и есть способ проверить, стало ли лучше.

Примеры, которые обычно получаются сильными:

- разбор обращений или заявок по категориям и приоритетам
- ассистент по внутренним документам компании или вуза
- помощник оператора: подсказывает ответ, человек подтверждает
- извлечение структурированных данных из писем, накладных, договоров
- проверка текстов на соответствие требованиям или регламенту

Примеры, которые обычно получаются слабыми: чат-бот общего назначения, «спроси что угодно по интернету», пересказ статей. У них нет измеримой задачи, и оценивать нечего.

## Правила игры

Это правила курса, на проекте они обязательны.

**1. Лестница вмешательства обязательна.** Занятие 45 начиналось с того, что способов повлиять на модель шесть, и дообучение среди них четвёртый, а не первый. В проекте вы обязаны сравнить **минимум три ступени** на одном и том же наборе:

- нулевая: базовый вызов модели без ухищрений
- первая: проработанный промпт, few-shot, структурированный вывод
- дальше ваша ступень по треку: retrieval, инструменты, дообучение, workflow

Работа, где сразу сделан агент или дообучение без сравнения с промптом, теряет баллы в двух критериях: и за методы адаптации, и за дисциплину замера.

**2. Оценочный набор фиксируется до начала работы.** Минимум тридцать примеров с эталонами. Собирается один раз, дальше не меняется. Все замеры на нём.

**3. Одно изменение за раз.** Поменяли промпт, модель и retrieval вместе, значит не узнали ничего.

**4. Числа сопровождаются неопределённостью.** На тридцати примерах разница в один ответ это три процента метрики. Занятие 46 было целиком про это: прежде чем объявлять улучшение, посчитайте интервал.

**5. Наблюдаемость не опциональна.** Каждый прогон должен быть виден в трекере: шаги, время, стоимость. Приложение без трейсинга это приложение, поведение которого вы не наблюдаете.

**6. Деньги считаются.** Стоимость одного обращения и стоимость обработки тысячи обращений. Это первый вопрос, который задаст заказчик.

**7. Ограничения называются вслух.** Работа, где всё получилось и ничего не сломалось, означает либо слишком простую задачу, либо неполный отчёт.

````python
# Служебный код проекта. Не меняйте эту ячейку, просто пользуйтесь функциями.

RESULTS = []

def add_result(name, score, step="", split="eval", note="", secs=None,
               cost_usd=None, **extra):
    '''Записать вариант в лидерборд.

    name     : как называется вариант, например "ступень 1: few-shot"
    score    : значение основной метрики
    step     : ступень лестницы: "0", "1", "2", "4", "6" или своя пометка
    note     : что именно изменено по сравнению с предыдущим вариантом
    secs     : среднее время одного обращения
    cost_usd : средняя стоимость одного обращения
    '''
    RESULTS.append({"name": name, "score": float(score), "step": str(step),
                    "split": split, "note": note, "secs": secs,
                    "cost_usd": cost_usd, **extra})
    print(f"[лидерборд] {name}: {score:.3f}" +
          (f" | {secs:.2f} сек" if secs else "") +
          (f" | {cost_usd:.5f} USD" if cost_usd else ""))


def show_leaderboard(higher_is_better=True):
    if not RESULTS:
        print("лидерборд пуст")
        return None
    df = pd.DataFrame(RESULTS).sort_values("score", ascending=not higher_is_better)
    cols = [c for c in ["step", "name", "score", "secs", "cost_usd", "split", "note"]
            if c in df.columns]
    display(df[cols].reset_index(drop=True))

    fig, ax = plt.subplots(figsize=(9, 0.55 * len(df) + 1.2))
    ax.barh(df["name"], df["score"], color="#1C4A6E")
    for i, v in enumerate(df["score"]):
        ax.text(v, i, f" {v:.3f}", va="center", fontsize=10)
    ax.invert_yaxis()
    ax.set_xlabel("основная метрика")
    ax.set_title("Лидерборд проекта")
    ax.grid(True, axis="x", color="#D9DEE3", lw=0.8)
    ax.set_axisbelow(True)
    for s in ("top", "right", "left"):
        ax.spines[s].set_visible(False)
    plt.tight_layout(); plt.show()
    return df


def data_fingerprint(items, key="input"):
    '''Отпечаток оценочного набора: шестнадцать символов от отсортированных входов.

    Нужен, чтобы доказать, что все замеры сделаны на одном и том же наборе.
    Если отпечаток изменился между замерами, сравнивать их нельзя.
    '''
    joined = "\n".join(sorted(str(it[key]) for it in items))
    return hashlib.sha256(joined.encode("utf-8")).hexdigest()[:16]


def evaluate(items, predict_fn, scorers, name, step="", note="", show_errors=5):
    '''Прогнать вариант по оценочному набору.

    items      : список словарей с ключами input и expected
    predict_fn : функция, принимает строку вопроса, возвращает строку ответа
    scorers    : словарь {имя: функция(item, output) -> 0.0 или 1.0}
    '''
    rows, t0 = [], time.time()
    for it in items:
        started = time.time()
        try:
            out = predict_fn(it["input"])
        except Exception as exc:
            out = f"[ОШИБКА: {type(exc).__name__}: {str(exc)[:80]}]"
        row = {"input": it["input"], "expected": it.get("expected"), "output": out,
               "sec": round(time.time() - started, 2)}
        for sname, fn in scorers.items():
            try:
                row[sname] = float(fn(it, out))
            except Exception:
                row[sname] = 0.0
        rows.append(row)
    total = time.time() - t0

    df = pd.DataFrame(rows)
    main = list(scorers)[0]
    score = df[main].mean()
    n = len(df)
    se = math.sqrt(max(score * (1 - score), 1e-9) / n)

    print(f"=== {name} ===")
    print(f"примеров: {n} | время всего: {total:.0f} сек | среднее на пример: {df['sec'].mean():.2f} сек")
    for sname in scorers:
        print(f"  {sname}: {df[sname].mean():.3f}")
    print(f"интервал 95 процентов для основной метрики: плюс-минус {1.96 * se:.3f}")
    print(f"шаг метрики (один пример): {1 / n:.3f}")

    if show_errors:
        bad = df[df[main] < 1.0].head(show_errors)
        if len(bad):
            print(f"\nпримеры ошибок ({len(df[df[main] < 1.0])} всего):")
            for _, r in bad.iterrows():
                print(f"  вопрос : {str(r['input'])[:90]}")
                print(f"  эталон : {str(r['expected'])[:90]}")
                print(f"  ответ  : {str(r['output'])[:120]}")
                print()

    add_result(name, score, step=step, note=note, secs=df["sec"].mean())
    return df


def compare_runs(df_a, df_b, metric, name_a="вариант A", name_b="вариант B"):
    '''Парное сравнение двух вариантов на одном наборе.

    Сравнивать два средних недостаточно: информация о разнице живёт
    в примерах, где варианты разошлись. Это критерий Макнемара.
    '''
    from scipy.stats import binomtest
    a, b = df_a[metric].values, df_b[metric].values
    assert len(a) == len(b), "варианты мерились на разном числе примеров, сравнение незаконно"

    both_right = int(((a == 1) & (b == 1)).sum())
    both_wrong = int(((a == 0) & (b == 0)).sum())
    only_a = int(((a == 1) & (b == 0)).sum())
    only_b = int(((a == 0) & (b == 1)).sum())

    print(f"{name_a} против {name_b}")
    print(f"  обе правы          : {both_right}")
    print(f"  обе неправы        : {both_wrong}")
    print(f"  прав только {name_a}: {only_a}")
    print(f"  прав только {name_b}: {only_b}")
    print(f"  информативных примеров: {only_a + only_b} из {len(a)}")

    if only_a + only_b == 0:
        print("  вердикт: варианты совпали на всех примерах, разницы нет")
        return 1.0
    p = min(binomtest(min(only_a, only_b), only_a + only_b, 0.5).pvalue * 2, 1.0)
    print(f"  вероятность случайности: {p:.3f}")
    print("  вердикт:", "различие устойчиво" if p < 0.05 else "перевес неотличим от монеты")
    return p


def cost_report(prompt_tokens, completion_tokens, price_in, price_out,
                requests_per_month=1000, usd_kzt=458.87):
    '''Стоимость одного обращения и месячная стоимость. Цены смотрите у провайдера.'''
    per_req = prompt_tokens / 1e6 * price_in + completion_tokens / 1e6 * price_out
    print(f"токенов на обращение: {prompt_tokens} вход, {completion_tokens} выход")
    print(f"одно обращение : {per_req:.5f} USD = {per_req * usd_kzt:.2f} тенге")
    print(f"{requests_per_month} обращений: {per_req * requests_per_month:.2f} USD = "
          f"{per_req * requests_per_month * usd_kzt:.0f} тенге")
    return per_req

print("готово: add_result, show_leaderboard, data_fingerprint, evaluate, compare_runs, cost_report")
````

Аналитик подразделения финансового мониторинга (AML) банка. Он получает список клиентов, уже связанных

приоритета нет, и объяснение «почему этот» живёт в голове аналитика.

**2. Что именно делает система и где она встраивается в его работу?**

В начале разбора новой сети аналитик открывает экран: сеть раскрашена по ролям (консолидатор, транзит,

задаёт ассистенту вопросы своими словами — «кого проверять первым», «кто собирает деньги с этих

открывается карточка узла. Система помогает, а не заменяет: решение о проверке и о сообщении в
регулятор принимает аналитик. Ассистент только читает анализ и ничего не блокирует и не отправляет.

[ВПИСАТЬ ОЦЕНКУ: сколько времени аналитик сейчас тратит на первичный разбор одной такой сети и

**4. Что будет, если система ошибётся?**

Ошибка дорогая в обе стороны. Пропущенное звено — непроверенный канал отмывания. Ложное обвинение —
юридический и репутационный риск для банка и вред для клиента. Отсюда архитектура: человек в контуре;
только чтение; выводы как гипотезы («признаки консолидации»), а не вердикты; отказ, когда данных нет;
каждое число в ответе обязано прийти из вызова инструмента, и каждый ответ виден трейсом.

Роли, кластеры и приоритеты считает не модель: это формальные правила с порогами и графовые метрики,

LLM нужна как интерфейс к графу: перевести свободный вопрос аналитика в последовательность запросов

сама по себе модель здесь бесполезна — без данных она не знает ни одного gid.

# 4. Данные и оценочный набор

Раздел на десять баллов, и с него начинается вся дальнейшая работа.

**Что должно быть в разделе:**

- источник данных или документов, объём, лицензия или разрешение на использование
- ответ про персональные данные: есть ли они, что с ними делаете

- отпечаток набора, зафиксированный кодом

# 5. Метрика и оценщики

Что именно вы считаете правильным ответом. Это нужно решить **до** первого замера, иначе метрика подстроится под результат.

Два типа оценщиков, нужны оба:

- **правило**: проверяет проверяемое. Есть ли в ответе нужный идентификатор, правильная ли категория, соблюдён ли формат, есть ли ссылка на источник
- **судья**: другая модель оценивает то, что правилом не выразить. Не выдумала ли система данные, ответила ли на языке вопроса, отказалась ли там, где надо было отказаться

Про судью помните: он ошибается и склонен одобрять ответы, похожие на свои. Используйте его как дешёвый сигнал на объёме, а не как истину. Хорошая практика: разметить руками десять примеров и посмотреть, насколько судья с вами совпал.

````python
# Оценщик-правило: проверяет то, что проверяемо, по полю check каждого примера (eval/README.md).
GID_RE = re.compile(r"(?<!\d)\d{15,20}(?!\d)")

ROLE_STEM = {
    "consolidator": "консолид", "transit": "транзит", "distributor": "распредел",
    "terminal": "конечн", "coordinator": "координ", "peripheral": "периф",
}

# Признаки отказа: ассистент говорит, что ответить или сделать не может, а не сочиняет.
REFUSAL_RE = re.compile(
    r"не могу|не можем|нет данных|нет доступа|не располага|не имею|отсутству|не найден|не существует|"
    r"нет информац|не удалось найти|нельзя|не выношу|не делаю вывод|не делаю выводов|вне того|не знаю|"
    r"не вижу|нет сведений|недоступ|не содержит|нет в анализе|в анализе нет|нет в графе|в графе нет|"
    r"нет такого|нет клиента|не входит|не позволя|не собира|не располож|не выполня|не блокир|не отправля|"
    r"только чита|не могу ответить|за пределами|вне данных|не охватыва|охватывает только|"
    r"(переводов|данных|сведений|информации)[^.]{0,40}\bнет\b",
    re.IGNORECASE,
)

_NUM_RE = re.compile(r"\d[\d   ]*(?:[.,]\d+)?(?:\s*(млрд|млн|тыс))?", re.IGNORECASE)


def numbers_in(text):
    """Все числа ответа, с «1,5 млн» → 1 500 000 и пробелами-разделителями тысяч."""
    out = []
    for m in _NUM_RE.finditer(text.replace(" ", " ").replace(" ", " ")):
        raw = m.group(0)
        unit = (m.group(1) or "").lower()
        digits = re.sub(r"(млрд|млн|тыс)", "", raw, flags=re.IGNORECASE).replace(" ", "").replace(",", ".")
        try:
            value = float(digits.rstrip("."))
        except ValueError:
            continue
        value *= {"млрд": 1e9, "млн": 1e6, "тыс": 1e3}.get(unit, 1)
        out.append(value)
    return out


def rule_score(item, output):
    c, text = item["check"], output or ""
    low = text.lower()
    if c.get("refusal"):
        invented = [g for g in GID_RE.findall(text) if g not in c.get("allowedGids", [])]
        return float(bool(REFUSAL_RE.search(text)) and not invented)
    # Gid, который аналитик сам назвал в вопросе, ответ повторять не обязан: «у него 34 получателя»
    # отвечает про того же клиента. Требуется только gid, которого в вопросе не было.
    named = set(GID_RE.findall(item["input"]))
    present = lambda g: g in text or g in named
    if "gidsAll" in c and not all(present(g) for g in c["gidsAll"]):
        return 0.0
    if "gidsAny" in c and not any(g in text for g in c["gidsAny"]):
        return 0.0
    if "role" in c and ROLE_STEM[c["role"]] not in low:
        return 0.0
    if "numbers" in c:
        found = numbers_in(text)
        if not all(any(abs(v - n["value"]) <= n["tol"] for v in found) for n in c["numbers"]):
            return 0.0
    if "mentionsAny" in c and not any(m in low for m in c["mentionsAny"]):
        return 0.0
    return 1.0


# Оценщик-судья: другая модель решает то, что правилом не выразить — выдумано ли что-то, есть ли
# вывод о виновности, честен ли отказ. Судья видит данные, которые были у ассистента (результаты
# инструментов), иначе любая верная подробность сверх короткого эталона выглядит для него выдумкой.
# Просим одну цифру, а не рассуждение.
JUDGE_PROMPT = '''Ты проверяешь ответ ассистента AML-аналитика. Ассистент отвечает по графу банковских
переводов за июль 2026 года и только читает анализ: он не может ничего блокировать, отправлять или менять.
Ответь ровно одной цифрой.

1 — ответ по существу согласуется с ОЖИДАНИЕМ: gid и числа из ОЖИДАНИЯ в ответе есть или ему не
противоречат (округление допустимо). Подробности сверх ожидания допустимы, если они подтверждаются
ДАННЫМИ АССИСТЕНТА. Если ОЖИДАНИЕ «отказ», ассистент честно говорит, что ответить или сделать этого не
может, и не сочиняет факты.
0 — ответ противоречит ОЖИДАНИЮ; конкретные gid, суммы или роли не подтверждаются ДАННЫМИ АССИСТЕНТА
(если данных нет, любые конкретные факты о клиентах выдуманы); ответ не по существу; вместо отказа
сочинены сведения; клиент назван преступником или виновным (допустимы только «признаки» и «требует проверки»).

ВОПРОС: {q}
ОЖИДАНИЕ: {e}
ДАННЫЕ АССИСТЕНТА (результаты его инструментов): {d}
ОТВЕТ АССИСТЕНТА: {o}

Только цифра.'''

EVIDENCE_LIMIT = 12000  # символов: карточка узла с соседями бывает длиннее 6 000


def evidence_for(question):
    """Результаты инструментов этого ответа, из последнего прогона. Нет инструментов — нет данных."""
    call = CALLS.get(CURRENT_VARIANT, {}).get(question)
    if not call or not call.get("toolCalls"):
        return "нет — ассистент не вызывал инструментов"
    parts = [f"{t['name']}({json.dumps(t['arguments'], ensure_ascii=False)}) → "
             f"{json.dumps(t['result'], ensure_ascii=False)}" for t in call["toolCalls"]]
    return "\n".join(parts)[:EVIDENCE_LIMIT]


def judge_score(item, output):
    r = oa.chat.completions.create(
        model=JUDGE_MODEL,
        messages=[{"role": "user", "content": JUDGE_PROMPT.format(
            q=item["input"], e=item.get("expected"), d=evidence_for(item["input"]), o=output)}],
    )
    raw = (r.choices[0].message.content or "").strip()
    return 1.0 if raw.startswith("1") else 0.0



SCORERS = {"правило": rule_score, "судья": judge_score}
print("оценщиков:", list(SCORERS), "| судья:", JUDGE_MODEL)
````

**Как правило и судья были откалиброваны — до первого официального замера.** На пилотном прогоне
ступени 2 (вне отчёта) правило отбраковало 5 верных ответов, где агент не повторил gid из вопроса
(«у него 34 получателя»), и один верный отказ («переводов за август нет, граф охватывает только июль»).
Правило поправлено: gid из вопроса повторять не обязательно, словарь отказов шире. Судья без данных
ставил 0 верным ответам, в которых подробностей больше, чем в коротком эталоне, — он не мог их
проверить и считал выдумкой. Теперь судья видит результаты инструментов ответа. Сам оценочный набор
не менялся: его отпечаток проверяется выше.

````python
def run_step(variant, name, step, note):
    """Ступень лестницы через evaluate из шаблона, плюс то, что вернул сервис: стоимость, токены, трейс."""
    global CURRENT_VARIANT
    CURRENT_VARIANT = variant
    df = evaluate(EVAL_SET, predictor(variant), SCORERS, name=name, step=step, note=note)
    calls = [CALLS.get(variant, {}).get(q, {}) for q in df["input"]]
    df["id"] = [it["id"] for it in EVAL_SET]
    df["kind"] = [it["kind"] for it in EVAL_SET]
    df["topic"] = [it["topic"] for it in EVAL_SET]
    df["trace_id"] = [c.get("traceId") for c in calls]
    df["cost_usd"] = [c.get("usage", {}).get("costUsd", np.nan) for c in calls]
    df["tokens_in"] = [c.get("usage", {}).get("inputTokens", np.nan) for c in calls]
    df["tokens_out"] = [c.get("usage", {}).get("outputTokens", np.nan) for c in calls]
    df["model_calls"] = [c.get("usage", {}).get("modelCalls", np.nan) for c in calls]
    df["server_ms"] = [c.get("latencyMs", np.nan) for c in calls]
    df["tools"] = [",".join(t["name"] for t in c.get("toolCalls", [])) for c in calls]
    RESULTS[-1]["cost_usd"] = float(df["cost_usd"].mean())
    RESULTS[-1]["judge"] = float(df["судья"].mean())
    by_kind = df.groupby("kind")[["правило", "судья"]].agg(["mean", "count"]).round(3)
    print("\nпо типам вопросов:")
    display(by_kind)
    print(f"стоимость ответа: {df['cost_usd'].mean():.6f} USD | токенов: {df['tokens_in'].mean():.0f} вход, "
          f"{df['tokens_out'].mean():.0f} выход | вызовов модели: {df['model_calls'].mean():.2f}")
    return df
````

# 8. Ваша ступень: агент с инструментами

**Что строим.** Продукт как он есть: тот же контракт и те же примеры, что на ступени 1, но источник
фактов — семь инструментов над готовым анализом, все только для чтения. Модель сама выбирает, что
вызвать и в каком порядке; цикл до 10 шагов.

| инструмент | что отвечает |
|---|---|
| `get_top_nodes` | кого проверять первым: ранг, роль, приоритет, причина; фильтр по роли или кластеру |
| `get_node` | карточка узла: роль с доказательством, метрики, 5 крупнейших отправителей и получателей |
| `get_cluster` | кластер: размер, seed, внутренний оборот, гипотеза, главные узлы |
| `find_collectors` | общие получатели денег нескольких узлов в пределах N переводов |
| `trace_flow` | куда уходят деньги от узла или откуда приходят |
| `simulate_removal` | что будет с сетью, если заблокировать узлы: компоненты до и после |
| `coverage_gaps` | чего данные не показывают и какой запрос отправить в банк |

**Параметры.** gpt-6-luna через Responses API; аргументы инструментов проверяются zod — это внешний ввод;
неизвестный gid — отказ с названием gid, а не пустой ответ; числа в результатах округлены до 2 знаков,
чтобы модель не цитировала 16 знаков; списки обрезаны и всегда несут общее число.

**Гипотеза.** Относительно ступени 1б вырастут вопросы о связях, конкретных узлах вне топ-50, удалении и

````python
# Сводка по ступеням: обе метрики с интервалом, по типам вопросов, стоимость и задержка сервера.
RUNS = {"0": df0, "1": df1, "1б": df1b, "2": df2, "2+": df3}


def ci(p, n):
    return 1.96 * math.sqrt(max(p * (1 - p), 1e-9) / n)


def wilson(p, n, z=1.96):
    # Интервал Уилсона. evaluate() печатает нормальное приближение «±», и на краю шкалы оно
    # вырождается: 40 из 40 дают «±0», будто ошибиться невозможно. Уилсон для 40/40 даёт 0.91–1.00.
    centre = (p + z * z / (2 * n)) / (1 + z * z / n)
    half = z * math.sqrt(p * (1 - p) / n + z * z / (4 * n * n)) / (1 + z * z / n)
    return max(0.0, centre - half), min(1.0, centre + half)


rows = []
for step, df in RUNS.items():
    n = len(df)
    p_rule, p_judge = df["правило"].mean(), df["судья"].mean()
    lo, hi = wilson(p_rule, n)
    jlo, jhi = wilson(p_judge, n)
    r = {"ступень": step, "правило": p_rule, "правило: ДИ Уилсона": f"{lo:.2f}–{hi:.2f}", "± правило": ci(p_rule, n),
         "судья": p_judge, "судья: ДИ Уилсона": f"{jlo:.2f}–{jhi:.2f}", "± судья": ci(p_judge, n),
         "_lo": lo, "_hi": hi, "_jlo": jlo, "_jhi": jhi}
    for kind in ["типовой", "пограничный", "отказ"]:
        part = df[df["kind"] == kind]
        r[f"{kind} ({len(part)})"] = part["правило"].mean()
    r["USD/ответ"] = df["cost_usd"].mean()
    r["токенов вход"] = df["tokens_in"].mean()
    r["задержка сервера, с"] = df["server_ms"].mean() / 1000
    r["задержка p90, с"] = df["server_ms"].quantile(0.9) / 1000
    rows.append(r)
SUMMARY = pd.DataFrame(rows).set_index("ступень")
display(SUMMARY.drop(columns=["_lo", "_hi", "_jlo", "_jhi"]).round(4))

fig, ax = plt.subplots(figsize=(8, 3.2))
x = np.arange(len(SUMMARY))
rule_err = [SUMMARY["правило"] - SUMMARY["_lo"], SUMMARY["_hi"] - SUMMARY["правило"]]
judge_err = [SUMMARY["судья"] - SUMMARY["_jlo"], SUMMARY["_jhi"] - SUMMARY["судья"]]
ax.bar(x - 0.18, SUMMARY["правило"], 0.36, yerr=rule_err, capsize=3, label="правило", color="#1C4A6E")
ax.bar(x + 0.18, SUMMARY["судья"], 0.36, yerr=judge_err, capsize=3, label="судья", color="#8FB3CC")
ax.set_xticks(x, SUMMARY.index)
ax.set_ylim(0, 1.05)
ax.set_ylabel("доля верных")
ax.set_title("Лестница: обе метрики, 95% интервал Уилсона")
ax.legend(frameon=False)
ax.grid(True, axis="y", color="#D9DEE3", lw=0.8)
ax.set_axisbelow(True)
for s_ in ("top", "right"):
    ax.spines[s_].set_visible(False)
plt.tight_layout(); plt.show()

print("согласие правила и судьи по ступеням:",
      {step: f"{(df['правило'] == df['судья']).mean():.2f}" for step, df in RUNS.items()})
````

# 11. Наблюдаемость

Раздел на десять баллов. Приложение без трейсинга это приложение, поведение которого вы не наблюдаете.

Что должно быть сделано:

- трейсинг подключён, все шаги видны: вызовы модели, инструменты, retrieval
- в трейсах проставлены метаданные: сессия, пользователь, теги окружения
- **связь с трекером проверена явно**: при неверных ключах SDK молчит и трейсы теряются без единой ошибки
- посчитаны стоимость и задержка по шагам
- на трейсы ставятся оценки: от правила, от судьи или от пользователя
- **разобрана хотя бы одна конкретная ошибка по трейсу**, с выводом, что именно её вызвало

Последний пункт самый ценный. Скриншот интерфейса это не наблюдаемость, наблюдаемость это когда вы по трейсу нашли причину.

````python
# Подключение трекера. Ключи только через getpass (или окружение), не строкой в коде.
for k in ["LANGFUSE_PUBLIC_KEY", "LANGFUSE_SECRET_KEY"]:
    if not os.environ.get(k):
        os.environ[k] = getpass.getpass(f"{k}: ")
os.environ.setdefault("LANGFUSE_HOST", "https://cloud.langfuse.com")


# Не SDK, а прямые запросы: для организаций Langfuse, созданных после 16.09.2026, старые API чтения
# (/api/public/traces, v1/v2 scores) отвечают 410, а старый эндпоинт записи молча теряет события.
# Сервис пишет трейсы по OpenTelemetry; здесь — чтение через /api/public/v2/observations и v3/scores.
class LangfuseREST:
    """Публичный API Langfuse. У облака есть лимит запросов: на 429 и 5xx — пауза и повтор
    (сколько просит Retry-After, иначе 2, 4, 8… секунд), а не падение ячейки. Первый прогон упал
    именно так: 160 оценок подряд, затем чтение трейсов — 429."""

    ATTEMPTS = 6

    def __init__(self):
        self.base = os.environ["LANGFUSE_HOST"].rstrip("/")
        self.auth = (os.environ["LANGFUSE_PUBLIC_KEY"], os.environ["LANGFUSE_SECRET_KEY"])
        self.retries = 0

    def _request(self, method, path, **kw):
        for attempt in range(self.ATTEMPTS):
            r = requests.request(method, self.base + path, auth=self.auth, timeout=30, **kw)
            if r.status_code != 429 and r.status_code < 500:
                return r
            if attempt == self.ATTEMPTS - 1:
                return r
            self.retries += 1
            wait = r.headers.get("Retry-After")
            time.sleep(min(float(wait), 60) if wait and wait.replace(".", "", 1).isdigit() else 2 ** (attempt + 1))
        return r

    def auth_check(self):
        return self._request("GET", "/api/public/projects").status_code == 200

    def observations(self, trace_id, hours=48):
        now = time.time()
        params = {"traceId": trace_id, "fields": "core,basic,usage,io",
                  "fromStartTime": time.strftime("%Y-%m-%dT%H:%M:%SZ", time.gmtime(now - hours * 3600)),
                  "toStartTime": time.strftime("%Y-%m-%dT%H:%M:%SZ", time.gmtime(now + 300))}
        r = self._request("GET", "/api/public/v2/observations", params=params)
        r.raise_for_status()
        return sorted(r.json().get("data", []), key=lambda o: o["startTime"])

    def scores(self, trace_id):
        r = self._request("GET", "/api/public/v3/scores",
                          params={"traceId": trace_id, "fields": "core,subject,details"})
        r.raise_for_status()
        return r.json().get("data", [])

    def score(self, trace_id, name, value, comment=None):
        r = self._request("POST", "/api/public/scores", json={
            "traceId": trace_id, "name": name, "value": float(value), "dataType": "NUMERIC",
            "comment": comment, "environment": "notebook"})
        return r.status_code < 300


lf = LangfuseREST()

try:
    ok = lf.auth_check()
    print("связь с трекером:", ok)
except Exception as e:
    ok = False
    print("СВЯЗИ НЕТ:", type(e).__name__, str(e)[:120])
assert ok, "без связи трейсы теряются молча, дальше идти бессмысленно"
````
